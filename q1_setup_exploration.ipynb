### 2. `output/q1_exploration.csv`
**Format:** CSV file
**Required Columns (exact names):** `column_name`, `mean`, `std`, `min`, `max`, `missing_count`
**Content:** One row per numeric column in the dataset
- `column_name`: Name of the numeric column
- `mean`: Mean value (float)
- `std`: Standard deviation (float)
- `min`: Minimum value (float)
- `max`: Maximum value (float)
- `missing_count`: Number of missing values (integer)

**Example:**
```csv
column_name,mean,std,min,max,missing_count
Water Temperature,15.23,5.12,0.5,28.7,2500
Air Temperature,18.45,8.23,-5.2,35.8,1500
Wind Speed,6.78,4.56,0.1,25.3,0
```

**Content:** At least 2 plots in a single figure (use subplots)
**Required plots:**
1. **Distribution plot:** Histogram or density plot of at least one numeric variable

**Requirements:**
- Clear axis labels (xlabel, ylabel)
- Title for each subplot
- Overall figure title (optional but recommended)

- Saved as PNG with sufficient resolution (dpi=150 or higher)

````python
from IPython.display import display, Markdown

df = pd.read_csv('data/beach_sensors.csv')
#shape
t1 = ("# Dataset Overview")
t2 = f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns"
#column info
t3 = "Column Summary:"
t4 = df.dtypes.to_string()
#date range
t5 = "Date Range:"
t6 = f"Start Date Value: {df["Measurement Timestamp"].min()} , End Date Value: {df["Measurement Timestamp"].max()}"
#missing values
t7 = "Missing Values:"
t8 = df.isnull().sum().to_string()

texts = [t1,t2,t3,t4,t5,t6,t7,t8]
texts = "\n\n".join(texts)

with open("output/q1_data_info.txt", "w") as f:
    f.write(texts)









````

File 2

````python
numeric_cols = df.select_dtypes(include=['number']).columns
dict_use = []
for col in numeric_cols:
    dict_use.append({
        "column_name": col,
        "mean": df[col].mean(),
        "std": df[col].std(),
        "min": df[col].min(),
        "max": df[col].max(),
        "missing_count": df[col].isnull().sum()
    })

summary = pd.DataFrame(dict_use)
summary.to_csv("output/q1_exploration.csv",index=False)

````

file 3

````python
fig, axes = plt.subplots(1, 2, figsize=(15, 12))
AT_distribution = df["Air Temperature"]
AT_distribution.plot(kind="hist",ax=axes[0],edgecolor="black")
axes[0].set_title("Distribution Plot of Air Temperature",fontweight="bold")
axes[0].set_xlabel('Air Temperature')
axes[0].set_ylabel('Count')


df["Measurement Timestamp"] = pd.to_datetime(df["Measurement Timestamp"])
df = df.set_index("Measurement Timestamp")
temp_daily = df["Rain Intensity"].resample("D").mean()
axes[1].plot(temp_daily.index, temp_daily.values ,color='forestgreen', linewidth=2)
axes[1].set_title("Time Series Plot of Rain Intensity",fontweight="bold")
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Rain Intensity')


plt.tight_layout()
plt.savefig('output/q1_visualizations.png', dpi=300, bbox_inches='tight')
plt.show()

````
